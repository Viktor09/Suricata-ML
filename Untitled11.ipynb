```python
class ARPSpoofingModel(nn.Module):
    def __init__(self,
                 static_dim=10,
                 dynamic_local_dim=9,
                 dynamic_global_dim=3,
                 dim_model=64,
                 num_head=8,
                 num_encoder_layers=3,
                 dim_feedforward=256,
                 dropout=0.1):
        super().__init__()

        self.embedding = nn.Linear(static_dim, dim_model)
        self.pos_encoder = nn.Parameter(torch.randn(1, 50, dim_model) * 0.02)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=dim_model,
            nhead=num_head,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True,
            activation='gelu'
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_encoder_layers)

        self.dynamic_proj = nn.Linear(dynamic_local_dim + dynamic_global_dim, dim_model)

        self.classifier = nn.Sequential(
            nn.Linear(dim_model * 2, 64),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )

    def forward(self, x_static, x_dynamic):
        x = self.embedding(x_static)
        x = x + self.pos_encoder[:, :x.size(1), :]
        x = self.transformer(x)
        x_seq = x.mean(dim=1)

        x_dyn = self.dynamic_proj(x_dynamic)

        x_combined = torch.cat([x_seq, x_dyn], dim=1)
        return self.classifier(x_combined)
```

```python
def extract_features_from_pcap(pcap_file):
    packets = []
    first_ts = None

    for pkt_data, pkt_metadata in RawPcapReader(pcap_file):
        pkt = Ether(pkt_data)
        if ARP in pkt:
            arp_layer = pkt[ARP]

            raw_ts = pkt_metadata.sec + (pkt_metadata.usec / 1000000.0)

            if first_ts is None:
                first_ts = raw_ts

            relative_ts = raw_ts - first_ts

            packets.append({
                'timestamp': relative_ts,
                'arp_code': arp_layer.op,
                'arp_src_mac': arp_layer.hwsrc,
                'arp_dst_mac': arp_layer.hwdst,
                'arp_src_ip': arp_layer.psrc,
                'arp_dst_ip': arp_layer.pdst,
                'is_gratuitous': int(arp_layer.psrc == arp_layer.pdst),
                'is_probe': int(arp_layer.op == 1 and arp_layer.psrc == "0.0.0.0"),
                'is_broadcast': int(pkt.dst == "ff:ff:ff:ff:ff:ff")
            })

    return pd.DataFrame(packets)
```

```python
def calculate_static_features(scenario_df, scenario):

  df = pd.DataFrame()

  df['timedelta'] = scenario_df['timestamp'].diff().fillna(0)
  df['arp_code'] = scenario_df['arp_code']
  df['is_gratuitous'] = scenario_df['is_gratuitous']
  df['is_broadcast'] = scenario_df['is_broadcast']

  ip_and_id = {'0.0.0.0': 0}
  current_id = 1
  for index, row in scenario_df.iterrows():
    ip = row['arp_src_ip']
    if ip not in ip_and_id:
        ip_and_id[ip] = current_id
        current_id += 1

  scenario_df['src_id_arp'] = scenario_df['arp_src_ip'].map(ip_and_id)
  df['src_id_arp'] = scenario_df['src_id_arp']

  for index, row in scenario_df.iterrows():
    ip = row['arp_dst_ip']
    if ip not in ip_and_id:
        ip_and_id[ip] = current_id
        current_id += 1

  scenario_df['dst_id_arp'] = scenario_df['arp_dst_ip'].map(ip_and_id)
  df['dst_id_arp'] = scenario_df['dst_id_arp']

  mac_and_id = {'00:00:00:00:00:00': 0}
  current_id = 1
  for index, row in scenario_df.iterrows():
    mac = row['arp_src_mac']
    if mac not in mac_and_id:
        mac_and_id[mac] = current_id
        current_id += 1

  scenario_df['src_id_mac'] = scenario_df['arp_src_mac'].map(mac_and_id)
  df['src_id_mac'] = scenario_df['src_id_mac']

  for index, row in scenario_df.iterrows():
    mac = row['arp_dst_mac']
    if mac not in mac_and_id:
        mac_and_id[mac] = current_id
        current_id += 1

  scenario_df['dst_id_mac'] = scenario_df['arp_dst_mac'].map(mac_and_id)
  df['dst_id_mac'] = scenario_df['dst_id_mac']

  seen_pairs = set()
  ip_mac_seen = []
  for _, row in scenario_df.iterrows():
      pair = (row['arp_src_ip'], row['arp_src_mac'])
      ip_mac_seen.append(1 if pair in seen_pairs else 0)
      seen_pairs.add(pair)
  df['ip_mac_seen_before'] = ip_mac_seen

  ip_gateway = "192.168.2.1"
  df['is_gateway'] = 0
  for index, row in scenario_df.iterrows():
    if row['arp_src_ip'] == ip_gateway:
      df.loc[index, 'is_gateway'] = 1

  df['label'] = 0
  if scenario == 'attack1_arp.pcap':
     df.loc[71:5805, 'label'] = 1
  elif scenario == 'attack2_arp.pcap':
     df.loc[73: , 'label'] = 1
  elif scenario == 'attack3_arp.pcap':
     df.loc[123:4966, 'label'] = 1
  elif scenario == 'attack4_arp.pcap':
     df.loc[57:5096, 'label'] = 1
  elif scenario == 'attack5_arp.pcap':
     df.loc[109:4808, 'label'] = 1
  elif scenario == 'attack6_arp.pcap':
     df.loc[93:4666, 'label'] = 1
  else:
     df['label'] = 0

  return df
```

```python
def calculate_dynamic_features_global(scenario_df):
    global_history = {
        'ip_to_macs': {},
        'mac_to_ips': {}
    }
    mac_ip_counts = {}  # {ip: {mac: count}} - evitam O(n²)
    results = []

    for index, row in scenario_df.iterrows():
        src_ip = row['arp_src_ip']
        src_mac = row['arp_src_mac']

        # Update global state
        if src_ip not in global_history['ip_to_macs']:
            global_history['ip_to_macs'][src_ip] = set()
        global_history['ip_to_macs'][src_ip].add(src_mac)

        if src_mac not in global_history['mac_to_ips']:
            global_history['mac_to_ips'][src_mac] = set()
        global_history['mac_to_ips'][src_mac].add(src_ip)

        # Update counts incremental
        if src_ip not in mac_ip_counts:
            mac_ip_counts[src_ip] = {}
        mac_ip_counts[src_ip][src_mac] = mac_ip_counts[src_ip].get(src_mac, 0) + 1

        # R(i)
        macs_for_ip = global_history['ip_to_macs'][src_ip]
        if len(macs_for_ip) > 1:
            counts = mac_ip_counts[src_ip]
            total = sum(counts.values())
            max_freq = max(counts.values())
            R_i = 1 - (max_freq / total)
        else:
            R_i = 0.0

        MD_i = len(global_history['ip_to_macs'][src_ip])
        ID_m = len(global_history['mac_to_ips'][src_mac])

        results.append({
            'R_i': R_i,
            'MD_i': MD_i,
            'ID_m': ID_m
        })

    return pd.DataFrame(results, index=scenario_df.index)
```

```python
def calculate_dynamic_features_window(window_df, delta=0.1):
    features = {}

    n_requests = (window_df['arp_code'] == 1).sum()
    n_replies = (window_df['arp_code'] == 2).sum()
    features['RRR'] = n_replies / (n_requests + 1e-9)

    requested_ips = set()
    orphan_replies = 0
    for _, row in window_df.iterrows():
        if row['arp_code'] == 1:
            requested_ips.add(row['arp_dst_ip'])
        elif row['arp_code'] == 2:
            if row['arp_dst_ip'] not in requested_ips:
                orphan_replies += 1
    features['URR'] = orphan_replies / (n_replies + 1e-9)

    timedeltas = window_df['timedelta']
    features['volatility'] = (timedeltas < delta).sum() / len(window_df)

    features['dt_mean'] = timedeltas.mean()
    features['dt_std'] = timedeltas.std()
    features['dt_cv'] = features['dt_std'] / (features['dt_mean'] + 1e-9)

    features['BR'] = window_df['is_broadcast'].sum() / len(window_df)

    gratuitous = window_df[window_df['is_gratuitous'] == 1]
    features['GRQ'] = (gratuitous['arp_code'] == 1).sum() / len(window_df)
    features['GRP'] = (gratuitous['arp_code'] == 2).sum() / len(window_df)

    return features
```

```python
def create_windows(df, window_size=50, stride=10):
    all_windows = []
    for start in range(0, len(df) - window_size + 1, stride):
        window = df.iloc[start:start + window_size]
        features = calculate_dynamic_features_window(window)
        features['label'] = 1 if window['label'].mean() > 0.3 else 0
        all_windows.append(features)
    return pd.DataFrame(all_windows)
```

```python
benign_files = ["normal_traffic_arp.pcap", "high_load_arp.pcap", "active_browsing_arp.pcap", "vrrp_arp.pcap", "aliassing_arp.pcap"]
attack_files = ["attack1_arp.pcap", "attack2_arp.pcap", "attack3_arp.pcap", "attack4_arp.pcap", "attack5_arp.pcap", "attack6_arp.pcap"]

```

```python
# ─── Prepare scenario ──────────────────────────────────────
def prepare_scenario(scenario_df, scenario_name, window_size=50, stride=10):
    static_df = calculate_static_features(scenario_df.copy(), scenario_name)
    global_df = calculate_dynamic_features_global(scenario_df)

    static_cols = ['timedelta', 'arp_code', 'is_gratuitous', 'is_broadcast',
                   'src_id_arp', 'dst_id_arp', 'src_id_mac', 'dst_id_mac',
                   'ip_mac_seen_before', 'is_gateway']

    static_sequences, dynamic_features, labels = [], [], []

    for start in range(0, len(static_df) - window_size + 1, stride):
        window_static = static_df.iloc[start:start + window_size]
        static_sequences.append(window_static[static_cols].values)

        # Pass a window of the original scenario_df, which contains raw IP addresses
        window_for_dynamic = scenario_df.iloc[start:start + window_size].copy()
        # Calculate timedelta for this window
        window_for_dynamic['timedelta'] = window_for_dynamic['timestamp'].diff().fillna(0)
        local_feats = calculate_dynamic_features_window(window_for_dynamic)

        window_global = global_df.iloc[start:start + window_size]
        dyn_vector = list(local_feats.values()) + [
            window_global['R_i'].mean(),
            window_global['MD_i'].mean(),
            window_global['ID_m'].mean()
        ]
        dynamic_features.append(dyn_vector)
        labels.append(1 if window_static['label'].mean() > 0.3 else 0)

    return static_sequences, dynamic_features, labels


# ─── Colectare date ────────────────────────────────────────
all_static, all_dynamic, all_labels = [], [], []

all_files = benign_files + attack_files
for pcap_file in all_files:
    scenario_df = extract_features_from_pcap(pcap_file)
    s, d, l = prepare_scenario(scenario_df, pcap_file)
    all_static.extend(s)
    all_dynamic.extend(d)
    all_labels.extend(l)

all_static = np.array(all_static, dtype=np.float32)
all_dynamic = np.array(all_dynamic, dtype=np.float32)
all_labels = np.array(all_labels, dtype=np.float32)

print(f"Total ferestre: {len(all_labels)}")
print(f"Attack: {all_labels.sum():.0f} | Benign: {(1-all_labels).sum():.0f}")
```

```python
# ─── Normalizare ───────────────────────────────────────────
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

n_windows, seq_len, n_static = all_static.shape
scaler_static = StandardScaler()
all_static = scaler_static.fit_transform(
    all_static.reshape(-1, n_static)
).reshape(n_windows, seq_len, n_static).astype(np.float32)

scaler_dynamic = StandardScaler()
all_dynamic = scaler_dynamic.fit_transform(all_dynamic).astype(np.float32)


# ─── Dataset & DataLoader ──────────────────────────────────
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split

class ARPDataset(Dataset):
    def __init__(self, static_seq, dynamic_feat, labels):
        self.static = torch.FloatTensor(static_seq)
        self.dynamic = torch.FloatTensor(dynamic_feat)
        self.labels = torch.FloatTensor(labels)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return self.static[idx], self.dynamic[idx], self.labels[idx]

X_s_train, X_s_test, X_d_train, X_d_test, y_train, y_test = train_test_split(
    all_static, all_dynamic, all_labels,
    test_size=0.2, random_state=42, stratify=all_labels
)

train_loader = DataLoader(ARPDataset(X_s_train, X_d_train, y_train), batch_size=32, shuffle=True)
test_loader = DataLoader(ARPDataset(X_s_test, X_d_test, y_test), batch_size=32)


# ─── Training ──────────────────────────────────────────────
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = ARPSpoofingModel().to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
criterion = nn.BCELoss()
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=3, factor=0.5)

for epoch in range(50):
    model.train()
    train_loss = 0
    for x_s, x_d, y in train_loader:
        x_s, x_d, y = x_s.to(device), x_d.to(device), y.to(device)
        optimizer.zero_grad()
        pred = torch.sigmoid(model(x_s, x_d)).squeeze()
        loss = criterion(pred, y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        train_loss += loss.item()

    model.eval()
    val_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for x_s, x_d, y in test_loader:
            x_s, x_d, y = x_s.to(device), x_d.to(device), y.to(device)
            pred = torch.sigmoid(model(x_s, x_d)).squeeze()
            val_loss += criterion(pred, y).item()
            correct += ((pred > 0.5) == y).sum().item()
            total += y.size(0)

    avg_train = train_loss / len(train_loader)
    avg_val = val_loss / len(test_loader)
    acc = correct / total
    scheduler.step(avg_val)

    if epoch % 5 == 0:
        print(f"Epoch {epoch:3d} | train: {avg_train:.4f} | val: {avg_val:.4f} | acc: {acc:.4f}")


# ─── Evaluare completa ─────────────────────────────────────
model.eval()
all_preds, all_true = [], []
with torch.no_grad():
    for x_s, x_d, y in test_loader:
        x_s, x_d = x_s.to(device), x_d.to(device)
        pred = torch.sigmoid(model(x_s, x_d)).squeeze()
        all_preds.extend((pred > 0.5).cpu().numpy())
        all_true.extend(y.numpy())

print("\nConfusion Matrix:")
print(confusion_matrix(all_true, all_preds))
print("\nClassification Report:")
print(classification_report(all_true, all_preds, target_names=['benign', 'attack']))


# ─── Salvare ───────────────────────────────────────────────
# import pickle
# torch.save(model.state_dict(), 'arp_model.pth')
# with open('scaler_static.pkl', 'wb') as f:
#     pickle.dump(scaler_static, f)
# with open('scaler_dynamic.pkl', 'wb') as f:
#     pickle.dump(scaler_dynamic, f)
```
